## 1. Binomial distribution

We have an experiment that follows a binomial distribution with parameters n=20, p=0.7.


First calculate probability of observing k>=12:


```python
n=20
p=0.7

s=0
for k in range(12,21):
    s+=math.factorial(n)/(math.factorial(k) * math.factorial(n - k))*p**k*(1-p)**(n-k)
s
```

Now plot probability mass function and cumulative density function:

```python
probability_mass_function=np.zeros(21)
cumulative_density_function=np.zeros(21)

for k in range(21):
    probability_mass_function[k]=math.factorial(n)/(math.factorial(k) * math.factorial(n - k))*p**k*(1-p)**(n-k)
    cumulative_density_function[k]=cumulative_density_function[k-1]+math.factorial(n)/(math.factorial(k) 
                                    * math.factorial(n - k))*p**k*(1-p)**(n-k)
    
probability_mass_function,cumulative_density_function
```

```python
k=np.arange(0,21)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(k,probability_mass_function)
axes[0].plot(12,probability_mass_function[12],"r.")

axes[0].set_xlabel("k")
axes[0].set_ylabel("P(Y=k)")
axes[0].grid()

axes[1].bar(k,cumulative_density_function)
axes[1].plot(12,cumulative_density_function[12],"r.")


axes[1].set_xlabel("k")
axes[1].set_ylabel(r"P(Y$\leq$k)")
axes[1].grid()

plt.show()
```

Plots tell us that probability for k<=12 is way smaller then for k>=12.

## Gaussian distribution

```python
n=1000000
mu1,mu2 = 1,3
sigma = 1


x=np.random.normal(mu1,sigma,n)
y=np.random.normal(mu2,sigma,n)

s=x+y
```

```python
X=np.linspace(s.min()-1, s.max() + 1,100)
Y=(1/(np.sqrt(2*np.pi*2*sigma**2))) * np.exp(-(X-(mu1+mu2))**2 / (2*2*sigma**2))

plt.hist(s,bins= np.arange(s.min()-1, s.max() + 1, 0.1),density=True,alpha=0.6,label=r"Histogramm of $s_i$")
plt.plot(X,Y,"r-.",label=r"$N(\mu_1+\mu_2,2\sigma^2)$")

plt.xlabel("s")

plt.grid()
plt.legend()
plt.show()
```
