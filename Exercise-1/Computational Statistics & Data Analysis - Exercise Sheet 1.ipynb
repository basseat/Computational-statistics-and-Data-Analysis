```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample=np.random.normal(1,3,1000)
```

Compute sample variance via two methods:

```python
x_hat=np.sum(sample)/len(sample)

svar1=np.sum((sample-x_hat)**2)/len(sample)
svar2=np.sum(sample**2-x_hat**2)/len(sample)

print(x_hat, svar1, svar2)
print(f"Mittelwert weicht um {1-x_hat} und Varianz um {9-svar1} ab")
```

With Random Variable X we can also define another RV: aX+b, mit a=3 und b=5.
Compute the sample variance analogue which corresponds to the expression Var(aX+b), and the expression a^2Var(X).

```python
a=3
b=5

sample_new=a*sample+b

x_hat_new=np.sum(sample_new)/len(sample_new)
svar_new=np.sum(sample_new**2-x_hat_new**2)/len(sample_new)

svar_new, a**2*3**2
```

```python
samples=np.zeros((10,1000))

for i in range(10):
    samples[i]=np.random.poisson(i+1,1000)
samples
```

```python
for i in range(10):
    plt.hist(samples[i],histtype="step", bins=np.arange(samples[i].min(), samples[i].max() + 2, 1), label=r"$\mu$={}".format(i+1))

plt.legend()
plt.show()
```

With increasing parameter mu, mean of distribution increases and distribution gets wider.
